```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
        2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

best_day, best_week = days[best_index], best_index // 7 + 1
worst_day, worst_week = days[worst_index], worst_index // 7 + 1

print(f"Total sales   : KES {total_sales:,}")
print(f"Average daily : KES {average_sales:,.2f}")
print(f"Highest day   : KES {highest_sales:,} ({best_day}, week {best_week})")
print(f"Lowest day    : KES {lowest_sales:,} ({worst_day}, week {worst_week})")
```

```python
def week_totals(sales):
    week1_total = sum(sales[:7])
    week2_total = sum(sales[7:])
    return week1_total, week2_total


week1_total, week2_total = week_totals(daily_sales)
growth_pct = (week2_total - week1_total) / week1_total * 100

print(f"Week 1 total : KES {week1_total:,}")
print(f"Week 2 total : KES {week2_total:,}")
print(f"Week-on-week growth: {growth_pct:+.2f}%")
```

```python
def classify_day(amount, average):
    """Classify a day's sales relative to the average.

    Returns "Strong" if amount is more than 20% above average,
    "Weak" if more than 20% below average, otherwise "Normal".
    """
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    return "Normal"


day_classes = [classify_day(amount, average_sales) for amount in daily_sales]

class_counts = {}
for label in day_classes:
    class_counts[label] = class_counts.get(label, 0) + 1

print(f"Strong threshold: above KES {average_sales * 1.2:,.2f}")
print(f"Weak threshold  : below KES {average_sales * 0.8:,.2f}\n")
for i in range(len(daily_sales)):
    print(f"Week {i // 7 + 1} {days[i]}: KES {daily_sales[i]:>6,}  -> {day_classes[i]}")

print("\nCounts:", class_counts)
```

```python
weekday_sales = []
weekend_sales = []
for day, amount in zip(days, daily_sales):
    if day in ("Sat", "Sun"):
        weekend_sales.append(amount)
    else:
        weekday_sales.append(amount)

weekday_avg = sum(weekday_sales) / len(weekday_sales)
weekend_avg = sum(weekend_sales) / len(weekend_sales)
weekend_premium_pct = (weekend_avg - weekday_avg) / weekday_avg * 100

print(f"Weekday average (Mon-Fri): KES {weekday_avg:,.2f}")
print(f"Weekend average (Sat-Sun): KES {weekend_avg:,.2f}")
print(f"Weekend sales are {weekend_premium_pct:.1f}% higher than weekday sales")
```

```python
def products_to_reorder(stock_data):
    """Return a list of product names whose quantity is at or below the reorder level."""
    return [name for name, info in stock_data.items() if info["qty"] <= info["reorder"]]


reorder_list = products_to_reorder(stock)

margins = {}
for product, info in stock.items():
    margins[product] = (info["price"] - info["cost"]) / info["price"] * 100

print("Products to reorder:", ", ".join(reorder_list))
print()
for product, info in stock.items():
    print(f"{product:<9} stock {info['qty']:>2} kg / reorder at {info['reorder']:>2} kg | margin {margins[product]:.1f}%")
```

```python
line = "=" * 52
print(line)
print("STALL PERFORMANCE REPORT - Wanjiku, Gikomba Market".center(52))
print(line)

print("\nSALES SUMMARY (14 days)")
print(f"  {'Total sales':<22}KES {total_sales:>10,}")
print(f"  {'Average daily sales':<22}KES {average_sales:>10,.2f}")
print(f"  {'Best day':<22}KES {highest_sales:>10,}  ({best_day}, week {best_week})")
print(f"  {'Worst day':<22}KES {lowest_sales:>10,}  ({worst_day}, week {worst_week})")

print("\nWEEKLY GROWTH")
print(f"  {'Week 1 total':<22}KES {week1_total:>10,}")
print(f"  {'Week 2 total':<22}KES {week2_total:>10,}")
print(f"  {'Week-on-week growth':<22}{growth_pct:>+14.2f}%")

print("\nDAY CLASSIFICATION")
for label in ("Strong", "Normal", "Weak"):
    print(f"  {label:<22}{class_counts.get(label, 0):>3} days")

print("\nWEEKDAY VS WEEKEND (average per day)")
print(f"  {'Weekday (Mon-Fri)':<22}KES {weekday_avg:>10,.2f}")
print(f"  {'Weekend (Sat-Sun)':<22}KES {weekend_avg:>10,.2f}")
print(f"  {'Weekend premium':<22}{weekend_premium_pct:>14.1f}%")

print("\nSTOCK AND MARGINS")
print(f"  {'Product':<10}{'Stock':>7}{'Reorder':>9}{'Margin':>9}  Status")
for product, info in stock.items():
    status = "REORDER" if product in reorder_list else "OK"
    print(f"  {product:<10}{info['qty']:>5}kg{info['reorder']:>7}kg{margins[product]:>8.1f}%  {status}")

print("\n" + line)
```

## Recommendations for Wanjiku

1. **Focus on weekends.** Saturday and Sunday average KES 4,900 per day against KES 2,685 on weekdays, which is 82.5% more. All four Strong days were weekend days, so keep full stock and your best selling hours there.
2. **Restock Tomatoes and Sukuma now.** Tomatoes (18 kg vs reorder level 25 kg) and Sukuma (9 kg vs 15 kg) are below their reorder levels. Order enough to carry you into Friday and the weekend, when sales are highest.
3. **Push Avocado.** Avocado has the highest margin at 42.9% (KES 60 profit per kg) and Sukuma follows at 40.0%, while Onions earn the lowest at 30.0%. Give Avocado the best display spot and watch it closely, since its 12 kg is only 2 kg above its reorder level of 10 kg.
4. **Boost the slow weekdays.** Six of the 14 days were Weak (below about KES 2,654), all on Mondays, Tuesdays or Thursdays, with Tuesday of week 1 the lowest at KES 1,950. Try a midweek offer, for example on Potatoes, where you hold 55 kg against a reorder level of 30 kg.
5. **The business is growing.** Week 2 sales were KES 24,400 against KES 22,050 in week 1, a rise of 10.66%. Keep recording daily to confirm that this trend holds over a longer period.
